```python
from pathlib import Path
import json, warnings

def auto_extract_benchmark():
    """
    Walk directory for '*episode_batch*.stats.json',
    group them by the agent folder, which is the parent, and return a dict.
    """
    stats_paths = list(Path("/content").rglob("*episode_batch*.stats.json"))
    if not stats_paths:
        raise FileNotFoundError("No episode_batch stats.json files found under /content")

    by_agent = {}
    for p in stats_paths:
        parts  = p.parts
        run_idx = next(i for i, seg in enumerate(parts) if seg.startswith("run_"))
        agent   = parts[run_idx - 1]

        # load JSON
        with open(p) as f:
            d = json.load(f)

        # grab whichever key is present
        rewards = (
            d.get("rewards")
            or d.get("returns")
            or d.get("episode_rewards")
            or []
        )
        steps = (
            d.get("lengths")
            or d.get("steps")
            or d.get("episode_lengths")
            or []
        )

        # accumulate
        entry = by_agent.setdefault(agent, {"rewards": [], "steps": []})
        entry["rewards"].extend(rewards)
        entry["steps"].extend(steps)

    # warn if any agent ended up with zero episodes
    for agent, data in by_agent.items():
        if not data["rewards"]:
            warnings.warn(f"[{agent}] found stats file but extracted 0 rewards")

    return by_agent

# run it and save JSON
benchmark_results = auto_extract_benchmark()
with open("benchmark_results.json", "w") as f:
    json.dump(benchmark_results, f)

print("Agents found:", list(benchmark_results.keys()))
print("Total episodes:", sum(len(v["rewards"]) for v in benchmark_results.values()))
```
